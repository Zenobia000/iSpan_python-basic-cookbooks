### 習題 2：成績等級判斷 ⭐

**題目重述**：
輸入分數，輸出等級：
- 90-100：A 優秀
- 80-89：B 良好
- 70-79：C 中等
- 60-69：D 及格
- 0-59：F 不及格
- 超出範圍：輸入錯誤

**知識點**：
- ✅ 範圍判斷
- ✅ 條件順序（從大到小）
- ✅ 輸入驗證

````python
# 解答
score = 85

if score < 0 or score > 100:
    grade = "輸入錯誤"
    message = "分數必須在 0-100 之間"
elif score >= 90:
    grade = "A"
    message = "優秀"
elif score >= 80:
    grade = "B"
    message = "良好"
elif score >= 70:
    grade = "C"
    message = "中等"
elif score >= 60:
    grade = "D"
    message = "及格"
else:
    grade = "F"
    message = "不及格"

print(f"分數：{score}，等級：{grade}，評語：{message}")
````

### 習題 3：奇偶數判斷 ⭐

**知識點**：
- ✅ 模除運算子 `%`

````python
# 解答方法 1：傳統 if-else
num = 7

if num % 2 == 0:
    result = "偶數"
else:
    result = "奇數"

print(f"{num} 是{result}")
````

````python
# 解答方法 2：三元運算子（更簡潔）
num = 7
result = "偶數" if num % 2 == 0 else "奇數"
print(f"{num} 是{result}")
````

**延伸思考**：

### 習題 5：閏年判斷 ⭐⭐

**題目重述**：
判斷一個年份是否為閏年

**閏年規則**：
1. 能被 4 整除 且 不能被 100 整除，**或**
2. 能被 400 整除

**知識點**：
- ✅ 複雜邏輯組合（`and`, `or`）
- ✅ 模除運算

````python
# 解答
year = 2024

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    result = "閏年"
else:
    result = "平年"

print(f"{year} 年是{result}")

# 測試案例
test_years = [2000, 1900, 2024, 2023]
for y in test_years:
    is_leap = (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)
    print(f"{y}: {'閏年' if is_leap else '平年'}")
````

**測試案例驗證**：
- 2000：閏年（能被 400 整除）✓
- 1900：平年（能被 100 整除但不能被 400 整除）✓
- 2024：閏年（能被 4 整除但不能被 100 整除）✓
- 2023：平年（不能被 4 整除）✓

### 習題 6：BMI 健康評估 ⭐⭐

**題目重述**：
計算 BMI 並評估健康狀況

**BMI 分類**：
- < 18.5：過輕
- 18.5-24：正常
- 24-27：過重
- ≥ 27：肥胖

**知識點**：
- ✅ 數學計算與條件判斷結合
- ✅ 範圍判斷

````python
# 解答
weight = 70  # 公斤
height = 1.75  # 公尺

bmi = weight / (height ** 2)

if bmi < 18.5:
    category = "過輕"
    suggestion = "建議增加營養攝取"
elif bmi < 24:
    category = "正常"
    suggestion = "請維持健康生活習慣"
elif bmi < 27:
    category = "過重"
    suggestion = "建議控制飲食並增加運動"
else:
    category = "肥胖"
    suggestion = "建議諮詢醫師或營養師"

print(f"身高：{height} m")
print(f"體重：{weight} kg")
print(f"BMI：{bmi:.2f}")
print(f"評估：{category}")
print(f"建議：{suggestion}")
````

**重點提示**：
- BMI = 體重(kg) / 身高(m)²
- 使用 `{bmi:.2f}` 格式化輸出保留 2 位小數
- 條件從小到大判斷（< 18.5 → < 24 → < 27 → else）

**知識點**：

### 習題 13：電費計算系統 🔴

**題目重述**：
台電夏季電費累進計算：
- 0-120 度：每度 2.1 元
- 121-330 度：每度 3.02 元
- 331-500 度：每度 4.39 元
- 501-700 度：每度 5.44 元
- 701 度以上：每度 6.16 元

**知識點**：
- ✅ 累進費率計算邏輯
- ✅ 多級距階梯計費
- ✅ 變數累加技巧

````python
# 解答：電費累進計算
usage = 350  # 用電度數

print(f"用電度數：{usage} 度")
print("電費明細：")

total_fee = 0
remaining = usage

# 第 1 級距：0-120 度
if remaining > 0:
    tier1 = min(remaining, 120)
    fee1 = tier1 * 2.1
    total_fee += fee1
    remaining -= tier1
    print(f"  第 1 級距 (0-120 度)：{tier1:.0f} 度 × 2.1 = {fee1:.1f} 元")

# 第 2 級距：121-330 度
if remaining > 0:
    tier2 = min(remaining, 210)  # 330 - 120 = 210
    fee2 = tier2 * 3.02
    total_fee += fee2
    remaining -= tier2
    print(f"  第 2 級距 (121-330 度)：{tier2:.0f} 度 × 3.02 = {fee2:.1f} 元")

# 第 3 級距：331-500 度
if remaining > 0:
    tier3 = min(remaining, 170)  # 500 - 330 = 170
    fee3 = tier3 * 4.39
    total_fee += fee3
    remaining -= tier3
    print(f"  第 3 級距 (331-500 度)：{tier3:.0f} 度 × 4.39 = {fee3:.1f} 元")

# 第 4 級距：501-700 度
if remaining > 0:
    tier4 = min(remaining, 200)  # 700 - 500 = 200
    fee4 = tier4 * 5.44
    total_fee += fee4
    remaining -= tier4
    print(f"  第 4 級距 (501-700 度)：{tier4:.0f} 度 × 5.44 = {fee4:.1f} 元")

# 第 5 級距：701 度以上
if remaining > 0:
    tier5 = remaining
    fee5 = tier5 * 6.16
    total_fee += fee5
    print(f"  第 5 級距 (701 度以上)：{tier5:.0f} 度 × 6.16 = {fee5:.1f} 元")

print(f"總電費：{total_fee:.1f} 元")
````

**解題關鍵**：
1. 使用 `remaining` 變數追蹤剩餘度數
2. 每次計算該級距的實際度數：`min(remaining, 級距上限)`
3. 累加費用到 `total_fee`
4. 扣除已計算的度數

**常見錯誤**：
- ❌ 直接用總度數乘以最高費率（應該分段計算）
- ❌ 級距範圍計算錯誤（例如第 2 級距是 210 度，不是 330 度）
- ❌ 忘記處理小數（電費通常取到小數第 1 位）

### 習題 14：簡易計算機 🔴

**題目重述**：
設計支援四則運算的計算機：
- 支援 +, -, *, / 四種運算
- 檢查除數不為零
- 處理無效運算符號

**知識點**：
- ✅ if-elif-else 多重條件
- ✅ 巢狀條件處理特殊情況
- ✅ 錯誤檢查與處理

````python
# 解答：簡易計算機
num1 = 10
operator = "/"
num2 = 3

print(f"計算：{num1} {operator} {num2}")

if operator == "+":
    result = num1 + num2
    print(f"結果：{result}")
elif operator == "-":
    result = num1 - num2
    print(f"結果：{result}")
elif operator == "*":
    result = num1 * num2
    print(f"結果：{result}")
elif operator == "/":
    if num2 == 0:
        print("錯誤：除數不能為零")
    else:
        result = num1 / num2
        print(f"結果：{result}")
else:
    print(f"錯誤：不支援的運算符號 '{operator}'")
````

````python
# 進階版：完整測試案例
def calculator(num1, operator, num2):
    """簡易計算機函式"""
    print(f"\n計算：{num1} {operator} {num2}")
    
    if operator == "+":
        result = num1 + num2
        print(f"結果：{result}")
    elif operator == "-":
        result = num1 - num2
        print(f"結果：{result}")
    elif operator == "*":
        result = num1 * num2
        print(f"結果：{result}")
    elif operator == "/":
        if num2 == 0:
            print("錯誤：除數不能為零")
        else:
            result = num1 / num2
            print(f"結果：{result}")
    else:
        print(f"錯誤：不支援的運算符號 '{operator}'")

# 測試所有情況
calculator(10, "+", 5)   # 加法
calculator(10, "-", 5)   # 減法
calculator(10, "*", 5)   # 乘法
calculator(10, "/", 5)   # 除法
calculator(10, "/", 0)   # 除以零
calculator(10, "%", 5)   # 無效運算符
````

**重點提示**：
- 除法運算使用**巢狀 if** 檢查除數
- 最後使用 `else` 捕捉無效運算符
- 條件順序：先判斷運算符類型，再處理特殊情況

**常見錯誤**：
- ❌ 忘記檢查除數為零
- ❌ 使用 `=` 而非 `==` 比較運算符
- ❌ 忘記處理無效運算符（缺少 else）

設計完整的停車費計算系統：

- ✅ 無條件進位處理
- ✅ 折扣計算

````python
# 解答：停車費計算
import math

# 輸入資料
hours = 5.5
vehicle_type = "car"
is_holiday = True
has_monthly_pass = False

# 無條件進位到整數小時
billing_hours = math.ceil(hours)

print("停車資訊：")
print(f"  車輛類型：{'汽車' if vehicle_type == 'car' else '機車'}")
print(f"  停車時間：{hours} 小時（計費 {billing_hours} 小時）")
print(f"  日期類型：{'假日' if is_holiday else '平日'}")
print(f"  月票：{'有' if has_monthly_pass else '無'}")
print("\n費用計算：")

# 計算費用
if vehicle_type == "car":
    if is_holiday:
        # 假日汽車費率
        first_2_hours = min(billing_hours, 2)
        remaining_hours = max(0, billing_hours - 2)
        
        first_2_fee = first_2_hours * 40
        remaining_fee = remaining_hours * 30
        
        print(f"  前 {first_2_hours} 小時：{first_2_hours} × 40 = {first_2_fee} 元")
        if remaining_hours > 0:
            print(f"  後 {remaining_hours} 小時：{remaining_hours} × 30 = {remaining_fee} 元")
    else:
        # 平日汽車費率
        first_2_hours = min(billing_hours, 2)
        remaining_hours = max(0, billing_hours - 2)
        
        first_2_fee = first_2_hours * 30
        remaining_fee = remaining_hours * 20
        
        print(f"  前 {first_2_hours} 小時：{first_2_hours} × 30 = {first_2_fee} 元")
        if remaining_hours > 0:
            print(f"  後 {remaining_hours} 小時：{remaining_hours} × 20 = {remaining_fee} 元")
    
    subtotal = first_2_fee + remaining_fee
    
elif vehicle_type == "motorcycle":
    # 機車費率（單一費率）
    if is_holiday:
        hourly_rate = 25
    else:
        hourly_rate = 20
    
    subtotal = billing_hours * hourly_rate
    print(f"  {billing_hours} 小時 × {hourly_rate} = {subtotal} 元")
else:
    print("  錯誤：未知的車輛類型")
    subtotal = 0

print(f"  小計：{subtotal} 元")

# 月票優惠
if has_monthly_pass:
    discount = subtotal * 0.2
    final_fee = subtotal * 0.8
    print(f"  月票優惠：-{discount:.0f} 元（8 折）")
else:
    final_fee = subtotal
    print("  月票優惠：無")

print(f"應繳費用：{final_fee:.0f} 元")
````

**解題關鍵**：
1. 使用 `math.ceil()` 無條件進位時數
2. 階梯計費：分成「前 2 小時」和「之後」兩部分
3. 多重條件：車型（car/motorcycle）× 日期（平日/假日）
4. 折扣計算：原價 × 0.8 = 8 折

**常見錯誤**：
- ❌ 忘記進位（5.5 小時應計為 6 小時）
- ❌ 階梯計費邏輯錯誤（前 2 小時、後續小時費率不同）
- ❌ 條件判斷順序混亂（應先判斷車型，再判斷日期）

### 習題 18：所得稅計算器 🔴

**題目重述**：
設計個人所得稅計算器：
- 稅率級距：
  - 0-540,000：免稅
  - 540,001-1,210,000：5%
  - 1,210,001-2,420,000：12%
  - 2,420,001-4,530,000：20%
  - 4,530,001 以上：30%
- 扣除額：
  - 標準扣除額：單身 120,000，夫妻 240,000
  - 免稅額：每人 92,000（本人 + 撫養人數）

**知識點**：
- ✅ 累進稅率計算
- ✅ 扣除額邏輯
- ✅ 多層條件判斷
- ✅ 數值計算與條件結合

````python
# 解答：所得稅計算器
annual_income = 800000
dependents = 2
married = True

print("所得稅試算：")
print(f"  年收入：{annual_income:,} 元")

# 計算扣除額
if married:
    standard_deduction = 240000
    deduction_type = "夫妻"
else:
    standard_deduction = 120000
    deduction_type = "單身"

print(f"  標準扣除額：{standard_deduction:,} 元（{deduction_type}）")

# 計算免稅額（本人 + 撫養人數）
total_persons = 1 + dependents
exemption = total_persons * 92000
print(f"  免稅額：{exemption:,} 元（{total_persons} 人 × 92,000）")

# 計算課稅所得
taxable_income = annual_income - standard_deduction - exemption
taxable_income = max(0, taxable_income)  # 不能為負數
print(f"  課稅所得：{taxable_income:,} 元")

print("\n稅額計算：")

# 累進稅率計算
if taxable_income <= 540000:
    tax = 0
    print("  免稅級距：0 元")
    print(f"  應納稅額：{tax:,} 元（課稅所得未超過 540,000）")
elif taxable_income <= 1210000:
    tax = (taxable_income - 540000) * 0.05
    print(f"  5% 級距：{taxable_income - 540000:,} × 5% = {tax:,.0f} 元")
    print(f"  應納稅額：{tax:,.0f} 元")
elif taxable_income <= 2420000:
    # 第一級距 + 第二級距
    tax1 = (1210000 - 540000) * 0.05
    tax2 = (taxable_income - 1210000) * 0.12
    tax = tax1 + tax2
    print(f"  5% 級距：{1210000 - 540000:,} × 5% = {tax1:,.0f} 元")
    print(f"  12% 級距：{taxable_income - 1210000:,} × 12% = {tax2:,.0f} 元")
    print(f"  應納稅額：{tax:,.0f} 元")
elif taxable_income <= 4530000:
    # 前三級距累加
    tax1 = (1210000 - 540000) * 0.05
    tax2 = (2420000 - 1210000) * 0.12
    tax3 = (taxable_income - 2420000) * 0.20
    tax = tax1 + tax2 + tax3
    print(f"  5% 級距：{tax1:,.0f} 元")
    print(f"  12% 級距：{tax2:,.0f} 元")
    print(f"  20% 級距：{taxable_income - 2420000:,} × 20% = {tax3:,.0f} 元")
    print(f"  應納稅額：{tax:,.0f} 元")
else:
    # 前四級距 + 最高級距
    tax1 = (1210000 - 540000) * 0.05
    tax2 = (2420000 - 1210000) * 0.12
    tax3 = (4530000 - 2420000) * 0.20
    tax4 = (taxable_income - 4530000) * 0.30
    tax = tax1 + tax2 + tax3 + tax4
    print(f"  5% 級距：{tax1:,.0f} 元")
    print(f"  12% 級距：{tax2:,.0f} 元")
    print(f"  20% 級距：{tax3:,.0f} 元")
    print(f"  30% 級距：{taxable_income - 4530000:,} × 30% = {tax4:,.0f} 元")
    print(f"  應納稅額：{tax:,.0f} 元")
````

**解題關鍵**：
1. 先計算「課稅所得」= 年收入 - 扣除額 - 免稅額
2. 累進稅率：分段計算，每段用不同稅率
3. 婚姻狀況影響扣除額（單身 120,000 vs 夫妻 240,000）
4. 課稅所得不能為負數（用 `max(0, taxable_income)`）

**重點提示**：
- 使用 `:,` 格式化千分位（例如 `1,234,567`）
- 累進稅率需要**分段累加**，不是直接用最高稅率
- 注意級距邊界值（540,000, 1,210,000, 2,420,000, 4,530,000）

**常見錯誤**：
- ❌ 直接用最高稅率計算（應該分段累進）
- ❌ 扣除額計算錯誤（忘記加上免稅額）
- ❌ 級距範圍計算錯誤

### 習題 19：智慧家居控制系統 🟣

**題目重述**：
設計智慧家居自動控制系統，根據多種感測器資料決定設備狀態：
- 冷氣控制（溫度 + 模式）
- 燈光控制（光線 + 時間 + 模式）
- 除濕機控制（濕度）
- 窗簾控制（溫度 + 光線 + 模式）

**知識點**：
- ✅ 超複雜巢狀條件
- ✅ 多重感測器整合
- ✅ 優先順序邏輯
- ✅ 狀態計算（耗電量）

````python
# 解答：智慧家居控制系統

# 環境感測器資料
indoor_temp = 28        # 室內溫度
outdoor_temp = 35       # 室外溫度
humidity = 70           # 濕度 (%)
light_level = 200       # 光線強度 (lux)

# 時間與狀態
current_hour = 14       # 24 小時制
someone_home = True     # 是否有人在家
sleep_mode = False      # 是否為睡眠模式
energy_save = False     # 是否開啟節能模式

print("智慧家居控制狀態：\n")
print("環境資訊：")
print(f"  室內溫度：{indoor_temp}°C")
print(f"  室外溫度：{outdoor_temp}°C")
print(f"  濕度：{humidity}%")
print(f"  光線：{light_level} lux")
print(f"  時間：{current_hour:02d}:00")
print(f"  有人在家：{'是' if someone_home else '否'}")
print(f"  睡眠模式：{'是' if sleep_mode else '否'}")
print(f"  節能模式：{'是' if energy_save else '否'}")

print("\n設備控制：")

# 1. 冷氣控制
device_count = 0  # 開啟設備數量（用於耗電計算）

if someone_home and indoor_temp > 26:
    ac_status = "開啟"
    if sleep_mode:
        ac_temp = 28
    elif energy_save:
        ac_temp = 27
    else:
        ac_temp = 25
    print(f"  冷氣：{ac_status}（設定 {ac_temp}°C）")
    device_count += 1
else:
    print("  冷氣：關閉")

# 2. 燈光控制
if sleep_mode:
    light_status = "微弱燈光"
    light_reason = "睡眠模式"
    device_count += 0.2  # 微弱燈光耗電較少
elif 23 <= current_hour or current_hour < 6:
    light_status = "夜燈模式"
    light_reason = "深夜時段"
    device_count += 0.3
elif light_level < 300 and someone_home:
    light_status = "開啟"
    light_reason = "光線不足"
    device_count += 1
else:
    light_status = "關閉"
    light_reason = "光線充足或無人"

print(f"  燈光：{light_status}（{light_reason}）")

# 3. 除濕機控制
if humidity > 65 and someone_home:
    dehumidifier_status = "開啟"
    dehumidifier_reason = "濕度過高"
    device_count += 1
else:
    dehumidifier_status = "關閉"
    dehumidifier_reason = "濕度正常或無人"

print(f"  除濕機：{dehumidifier_status}（{dehumidifier_reason}）")

# 4. 窗簾控制
if sleep_mode:
    curtain_status = "關閉"
    curtain_reason = "睡眠模式"
elif outdoor_temp > 32 and light_level > 500:
    curtain_status = "關閉"
    curtain_reason = "室外溫度過高且光線強烈"
else:
    curtain_status = "開啟"
    curtain_reason = "環境適宜"

print(f"  窗簾：{curtain_status}（{curtain_reason}）")

# 5. 計算預估耗電量
if device_count >= 3:
    power_consumption = "高"
elif device_count >= 1.5:
    power_consumption = "中等"
else:
    power_consumption = "低"

print(f"\n預估耗電：{power_consumption}（{device_count:.1f} 個設備運行）")
````

**解題關鍵**：
1. 優先順序：睡眠模式 > 節能模式 > 一般模式
2. 條件組合：多個感測器資料同時判斷
3. 狀態追蹤：使用 `device_count` 計算開啟設備數
4. 清晰輸出：每個設備都說明開啟/關閉原因

**技巧提示**：
- 使用變數累加追蹤狀態（`device_count`）
- 微弱燈光計為 0.2 個設備（耗電較少）
- 深夜時段判斷：`23 <= current_hour or current_hour < 6`

### 習題 20：投資組合風險評估 🟣

**題目重述**：
設計投資組合風險評估系統：
- 年齡係數（積極型/穩健型/保守型）
- 財務穩定度評估（負債比 + 緊急基金）
- 股票配置比例計算（多層調整）
- 風險等級與預期報酬

**知識點**：
- ✅ 多層次計算邏輯
- ✅ 複雜條件調整機制
- ✅ 邊界值檢查
- ✅ 綜合評估輸出

````python
# 解答：投資組合風險評估

# 個人資料
age = 35
annual_income = 1200000      # 年收入
investment_experience = 5    # 投資經驗（年）
risk_tolerance = "medium"    # 風險承受度：low/medium/high

# 財務狀況
total_assets = 5000000       # 總資產
debt_ratio = 0.3             # 負債比例
emergency_fund = 600000      # 緊急準備金

# 投資目標
investment_period = 10       # 投資期間（年）
target_return = 0.08         # 目標報酬率
has_dependents = True        # 是否有撫養責任

print("投資風險評估報告：\n")

# 1. 基本資料
print("基本資料：")
print(f"  年齡：{age} 歲")
print(f"  年收入：{annual_income:,} 元")
print(f"  投資經驗：{investment_experience} 年")
risk_tolerance_name = {"low": "低", "medium": "中等", "high": "高"}
print(f"  風險承受度：{risk_tolerance_name.get(risk_tolerance, '未知')}")

# 2. 財務分析
print("\n財務分析：")
print(f"  總資產：{total_assets:,} 元")
print(f"  負債比例：{debt_ratio * 100:.0f}%")

# 計算緊急基金月數
emergency_months = emergency_fund / (annual_income / 12)
print(f"  緊急準備金：{emergency_fund:,} 元（年收入的 {emergency_months:.1f} 個月）")

# 評估財務穩定度
if debt_ratio < 0.3 and emergency_fund >= annual_income * 0.5:
    financial_stability = "穩定"
elif debt_ratio < 0.5 and emergency_fund >= annual_income * 0.25:
    financial_stability = "一般"
else:
    financial_stability = "不穩定"

print(f"  財務穩定度：{financial_stability}")

# 3. 投資建議
print("\n投資建議：")

# 基礎股票配置（100 - 年齡，上限 80%）
base_stock_ratio = min(100 - age, 80)
print(f"  基礎股票配置：{base_stock_ratio}%（100 - {age}）")

# 根據風險承受度調整
if risk_tolerance == "low":
    risk_adjustment = -20
elif risk_tolerance == "medium":
    risk_adjustment = 0
elif risk_tolerance == "high":
    risk_adjustment = 10
else:
    risk_adjustment = 0

if risk_adjustment != 0:
    print(f"  風險承受度調整：{risk_adjustment:+d}%（{risk_tolerance_name.get(risk_tolerance)}）")
else:
    print(f"  風險承受度調整：{risk_adjustment}%（{risk_tolerance_name.get(risk_tolerance)}）")

# 根據財務穩定度調整
if financial_stability == "穩定":
    stability_adjustment = 5
elif financial_stability == "一般":
    stability_adjustment = 0
else:
    stability_adjustment = -15

print(f"  財務穩定度調整：{stability_adjustment:+d}%（{financial_stability}）")

# 撫養責任調整
if has_dependents:
    dependent_adjustment = -10
    print(f"  撫養責任調整：{dependent_adjustment}%（有撫養責任）")
else:
    dependent_adjustment = 0
    print(f"  撫養責任調整：{dependent_adjustment}%（無撫養責任）")

# 計算最終股票配置（限制在 10%-90%）
final_stock_ratio = base_stock_ratio + risk_adjustment + stability_adjustment + dependent_adjustment
final_stock_ratio = max(10, min(90, final_stock_ratio))
final_bond_ratio = 100 - final_stock_ratio

print(f"  建議股票比例：{final_stock_ratio}%")
print(f"  建議債券比例：{final_bond_ratio}%")

# 4. 風險等級評估
print()
if final_stock_ratio >= 70:
    risk_level = "高風險"
    expected_return = "8-12%"
elif final_stock_ratio >= 50:
    risk_level = "中高風險"
    expected_return = "6-10%"
elif final_stock_ratio >= 30:
    risk_level = "中低風險"
    expected_return = "4-8%"
else:
    risk_level = "低風險"
    expected_return = "2-6%"

print(f"風險等級：{risk_level}")
print(f"預期年化報酬：{expected_return}")
````

**解題關鍵**：
1. 基礎配置：`min(100 - age, 80)` 限制最高 80%
2. 多層調整：風險承受度 → 財務穩定度 → 撫養責任
3. 邊界限制：`max(10, min(90, ratio))` 限制在 10%-90%
4. 綜合評估：根據最終股票比例判定風險等級

**重點提示**：
- 使用字典查找風險承受度名稱：`{"low": "低", ...}`
- 格式化輸出：`:+d` 顯示正負號（例如 `+5%`）
- 緊急基金月數計算：`emergency_fund / (annual_income / 12)`

**常見錯誤**：
- ❌ 忘記邊界限制（股票比例可能超過 100% 或低於 0%）
- ❌ 調整順序錯誤（應該累加所有調整後再限制範圍）
- ❌ 風險等級判斷邏輯錯誤（條件順序很重要）

## 💡 常見問題與除錯技巧

### Q1: 為什麼條件不成立？
**A**: 檢查：
1. 是否使用 `==` 而非 `=`
2. 變數型態是否正確（字串 vs 數字）
3. 條件順序是否正確

### Q2: 縮排錯誤怎麼辦？
**A**: 
1. 確保使用 4 個空格（不要用 Tab）
2. 同一區塊的程式碼必須有相同縮排
3. 使用編輯器的「顯示空白字元」功能

### Q3: 如何簡化複雜條件？
**A**:
1. 使用括號明確優先順序
2. 將複雜條件拆分成多個變數
3. 使用 De Morgan's Law 簡化 not 運算
