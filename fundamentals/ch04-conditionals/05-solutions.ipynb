### 習題 1：年齡分類系統 ⭐

**題目重述**：
設計一個年齡分類系統：
- 0-12 歲：兒童
- 13-19 歲：青少年
- 20-64 歲：成年人
- 65 歲以上：長者
- 負數：輸入錯誤

**知識點**：
- ✅ if-elif-else 多重條件
- ✅ 範圍判斷
- ✅ 輸入驗證

````python
# 解答
age = 25

if age < 0:
    category = "輸入錯誤"
elif age <= 12:
    category = "兒童"
elif age <= 19:
    category = "青少年"
elif age <= 64:
    category = "成年人"
else:
    category = "長者"

print(f"年齡：{age} 歲，分類：{category}")
````

**常見錯誤**：
- ❌ 條件順序錯誤（例如先判斷 `age <= 64`）
- ❌ 忘記處理負數
- ❌ 邊界值判斷錯誤（12, 19, 64）

### 習題 2：成績等級判斷 ⭐

**題目重述**：
輸入分數，輸出等級：
- 90-100：A 優秀
- 80-89：B 良好
- 70-79：C 中等
- 60-69：D 及格
- 0-59：F 不及格
- 超出範圍：輸入錯誤

**知識點**：
- ✅ 範圍判斷
- ✅ 條件順序（從大到小）
- ✅ 輸入驗證

````python
# 解答
score = 85

if score < 0 or score > 100:
    grade = "輸入錯誤"
    message = "分數必須在 0-100 之間"
elif score >= 90:
    grade = "A"
    message = "優秀"
elif score >= 80:
    grade = "B"
    message = "良好"
elif score >= 70:
    grade = "C"
    message = "中等"
elif score >= 60:
    grade = "D"
    message = "及格"
else:
    grade = "F"
    message = "不及格"

print(f"分數：{score}，等級：{grade}，評語：{message}")
````

**重點提示**：
- 先檢查輸入範圍（0-100）
- 從大到小判斷，確保邏輯正確
- `score >= 80` 隱含「且 < 90」（因為已排除 >= 90）

### 習題 3：奇偶數判斷 ⭐

**題目重述**：
判斷一個整數是奇數還是偶數

**知識點**：
- ✅ 模除運算子 `%`
- ✅ if-else 二選一邏輯

````python
# 解答方法 1：傳統 if-else
num = 7

if num % 2 == 0:
    result = "偶數"
else:
    result = "奇數"

print(f"{num} 是{result}")
````

````python
# 解答方法 2：三元運算子（更簡潔）
num = 7
result = "偶數" if num % 2 == 0 else "奇數"
print(f"{num} 是{result}")
````

**延伸思考**：
- `num % 2 == 1` vs `num % 2 != 0`：都可以判斷奇數
- 負數也適用（-7 % 2 = 1）

### 習題 4：三數最大值 ⭐⭐

**題目重述**：
找出三個數字中的最大值（不使用 `max()` 函式）

**知識點**：
- ✅ 多重條件判斷
- ✅ 邏輯運算子 `and`

````python
# 解答
a, b, c = 15, 28, 22

if a >= b and a >= c:
    maximum = a
elif b >= a and b >= c:
    maximum = b
else:
    maximum = c

print(f"三個數 {a}, {b}, {c} 的最大值是 {maximum}")
````

**其他解法**（巢狀條件）：
```python
if a >= b:
    if a >= c:
        maximum = a
    else:
        maximum = c
else:
    if b >= c:
        maximum = b
    else:
        maximum = c
```

### 習題 5：閏年判斷 ⭐⭐

**題目重述**：
判斷一個年份是否為閏年

**閏年規則**：
1. 能被 4 整除 且 不能被 100 整除，**或**
2. 能被 400 整除

**知識點**：
- ✅ 複雜邏輯組合（`and`, `or`）
- ✅ 模除運算

````python
# 解答
year = 2024

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    result = "閏年"
else:
    result = "平年"

print(f"{year} 年是{result}")

# 測試案例
test_years = [2000, 1900, 2024, 2023]
for y in test_years:
    is_leap = (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)
    print(f"{y}: {'閏年' if is_leap else '平年'}")
````

**測試案例驗證**：
- 2000：閏年（能被 400 整除）✓
- 1900：平年（能被 100 整除但不能被 400 整除）✓
- 2024：閏年（能被 4 整除但不能被 100 整除）✓
- 2023：平年（不能被 4 整除）✓

### 習題 6：BMI 健康評估 ⭐⭐

**題目重述**：
計算 BMI 並評估健康狀況

**BMI 分類**：
- < 18.5：過輕
- 18.5-24：正常
- 24-27：過重
- ≥ 27：肥胖

**知識點**：
- ✅ 數學計算與條件判斷結合
- ✅ 範圍判斷

````python
# 解答
weight = 70  # 公斤
height = 1.75  # 公尺

bmi = weight / (height ** 2)

if bmi < 18.5:
    category = "過輕"
    suggestion = "建議增加營養攝取"
elif bmi < 24:
    category = "正常"
    suggestion = "請維持健康生活習慣"
elif bmi < 27:
    category = "過重"
    suggestion = "建議控制飲食並增加運動"
else:
    category = "肥胖"
    suggestion = "建議諮詢醫師或營養師"

print(f"身高：{height} m")
print(f"體重：{weight} kg")
print(f"BMI：{bmi:.2f}")
print(f"評估：{category}")
print(f"建議：{suggestion}")
````

**重點提示**：
- BMI = 體重(kg) / 身高(m)²
- 使用 `{bmi:.2f}` 格式化輸出保留 2 位小數
- 條件從小到大判斷（< 18.5 → < 24 → < 27 → else）

### 習題 7：登入驗證系統 ⭐⭐

**題目重述**：
驗證使用者名稱和密碼

**知識點**：
- ✅ 字串比較
- ✅ 邏輯運算子 `and`
- ✅ 多重條件

````python
# 解答
correct_username = "admin"
correct_password = "python123"

username = input("使用者名稱：")
password = input("密碼：")

if username == correct_username and password == correct_password:
    print("✅ 登入成功！")
    print(f"歡迎，{username}")
elif username != correct_username and password != correct_password:
    print("❌ 登入失敗：使用者名稱和密碼都錯誤")
elif username != correct_username:
    print("❌ 登入失敗：使用者名稱錯誤")
else:
    print("❌ 登入失敗：密碼錯誤")
````

**安全性提示**：
- 實際應用中不應該明確告知哪個欄位錯誤（避免洩漏資訊）
- 密碼應該加密儲存
- 應該限制登入嘗試次數

### 習題 8：溫度單位轉換判斷 ⭐⭐

**題目重述**：
根據使用者選擇的單位進行溫度轉換

**轉換公式**：
- 攝氏轉華氏：F = C × 9/5 + 32
- 華氏轉攝氏：C = (F - 32) × 5/9

**知識點**：
- ✅ 使用者輸入處理
- ✅ 條件判斷
- ✅ 數學運算

````python
# 解答
print("溫度轉換程式")
print("1. 攝氏轉華氏")
print("2. 華氏轉攝氏")

choice = input("請選擇（1 或 2）：")
temperature = float(input("請輸入溫度："))

if choice == "1":
    result = temperature * 9/5 + 32
    print(f"{temperature}°C = {result:.2f}°F")
elif choice == "2":
    result = (temperature - 32) * 5/9
    print(f"{temperature}°F = {result:.2f}°C")
else:
    print("❌ 輸入錯誤，請選擇 1 或 2")
````

**常見錯誤**：
- ❌ 忘記處理無效輸入（非 1 或 2）
- ❌ 公式錯誤（運算順序）
- ❌ 忘記型態轉換（input 返回字串）

### 習題 9：四則運算計算機 ⭐⭐⭐

**題目重述**：
設計簡易計算機，支援 +、-、×、÷ 四種運算

**知識點**：
- ✅ 字串比較
- ✅ 多重條件
- ✅ 除零檢查

````python
# 解答
num1 = float(input("第一個數字："))
operator = input("運算子（+, -, *, /）：")
num2 = float(input("第二個數字："))

if operator == "+":
    result = num1 + num2
    print(f"{num1} + {num2} = {result}")
elif operator == "-":
    result = num1 - num2
    print(f"{num1} - {num2} = {result}")
elif operator == "*":
    result = num1 * num2
    print(f"{num1} × {num2} = {result}")
elif operator == "/":
    if num2 == 0:
        print("❌ 錯誤：除數不能為零")
    else:
        result = num1 / num2
        print(f"{num1} ÷ {num2} = {result:.2f}")
else:
    print("❌ 錯誤：無效的運算子")
````

**重點提示**：
- 除法前必須檢查除數是否為零
- 使用巢狀條件處理特殊情況
- 處理無效運算子輸入

### 習題 10：電影分級系統 ⭐⭐⭐

**題目重述**：
根據年齡和是否有家長陪同，判斷是否可以觀看電影

**分級規則**：
- G (普遍級)：所有人
- PG (保護級)：12 歲以上或有家長陪同
- R (限制級)：18 歲以上

**知識點**：
- ✅ 巢狀條件
- ✅ 多重邏輯判斷

````python
# 解答
age = int(input("您的年齡："))
rating = input("電影分級（G, PG, R）：").upper()
with_parent = input("是否有家長陪同？(Y/N)：").upper() == "Y"

if rating == "G":
    print("✅ 普遍級，所有人都可觀看")
elif rating == "PG":
    if age >= 12:
        print("✅ 保護級，12 歲以上可觀看")
    elif with_parent:
        print("✅ 保護級，未滿 12 歲需家長陪同，已有家長陪同")
    else:
        print("❌ 保護級，未滿 12 歲需家長陪同")
elif rating == "R":
    if age >= 18:
        print("✅ 限制級，18 歲以上可觀看")
    else:
        print("❌ 限制級，未滿 18 歲不得觀看")
else:
    print("❌ 錯誤：無效的分級")
````

**技巧提示**：
- 使用 `.upper()` 統一大小寫
- 巢狀條件處理複雜邏輯
- 清楚的輸出訊息

### 習題 15：綜合成績系統 ⭐⭐⭐⭐

**題目重述**：
整合成績、出席率、作業繳交，綜合評估學生表現

**評估規則**：
1. 出席率 < 70%：直接不及格
2. 成績 + 作業加分
3. 綜合評語

**知識點**：
- ✅ 複雜條件組合
- ✅ 多層判斷
- ✅ 條件內修改變數

````python
# 解答（完整版參考 lecture 範例 8）
student_name = "王小明"
score = 88
attendance_rate = 0.85
has_homework = True

print(f"===== {student_name} 的成績評估 =====")
print(f"考試分數：{score}")
print(f"出席率：{attendance_rate * 100:.0f}%")
print(f"作業繳交：{'是' if has_homework else '否'}")
print("\n--- 評估結果 ---")

# 基本成績評級
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

# 出席率檢查
if attendance_rate < 0.7:
    print("⚠️ 警告：出席率不足 70%")
    grade = "F"

# 作業加分
if has_homework and score >= 60:
    bonus = 5
    final_score = min(score + bonus, 100)
    print(f"✅ 作業加分：+{bonus} 分")
else:
    final_score = score

print(f"\n最終成績：{grade} ({final_score} 分)")
````

**學習重點**：
- 多層條件判斷的組織方式
- 條件內修改變數值
- 使用 `min()` 限制最高分數
- 清晰的輸出格式

### Q1: 為什麼條件不成立？

### Q2: 縮排錯誤怎麼辦？
**A**: 
1. 確保使用 4 個空格（不要用 Tab）
2. 同一區塊的程式碼必須有相同縮排
3. 使用編輯器的「顯示空白字元」功能
