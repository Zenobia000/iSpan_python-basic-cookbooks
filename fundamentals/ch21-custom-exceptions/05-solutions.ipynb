### 習題 1：NegativeNumberError（10 分）

**題目重述**：建立 `NegativeNumberError` 例外類別，當數值為負數時拋出。

**知識點**：
- ✅ 自訂例外繼承
- ✅ `__init__` 中呼叫 `super().__init__()`
- ✅ 儲存額外屬性
- ✅ 在函式中使用 `raise` 拋出例外

````python
# 習題 1 完整解答

class NegativeNumberError(ValueError):
    """負數錯誤例外"""
    def __init__(self, value):
        self.value = value
        message = f"負數不被允許：{value}"
        super().__init__(message)

def sqrt(x):
    """計算平方根"""
    if x < 0:
        raise NegativeNumberError(x)
    return x ** 0.5

# 測試 1：正常情況
print(f"sqrt(16) = {sqrt(16)}")  # 4.0

# 測試 2：負數
try:
    result = sqrt(-9)
    print(f"sqrt(-9) = {result}")
except NegativeNumberError as e:
    print(f"捕獲例外：{e}")
    print(f"值：{e.value}")
````

**重點提示**：
1. 繼承 `ValueError` 而非 `Exception`，因為這是「值」的錯誤
2. 使用 `super().__init__(message)` 設定例外訊息
3. 將 `value` 儲存為屬性，方便後續存取

**常見錯誤**：
- ❌ 忘記呼叫 `super().__init__()`
- ❌ 直接繼承 `Exception` 而非 `ValueError`
- ❌ 沒有儲存 `value` 屬性

### 習題 2：InvalidCredentialsError（10 分）

**題目重述**：建立 `InvalidCredentialsError` 例外，用於登入驗證失敗。

**知識點**：
- ✅ 根據條件產生不同錯誤訊息
- ✅ 計算派生屬性（`is_locked`）
- ✅ 使用 f-string 格式化訊息

````python
# 習題 2 完整解答

class InvalidCredentialsError(Exception):
    """登入驗證失敗例外"""
    def __init__(self, username, attempt_count):
        self.username = username
        self.attempt_count = attempt_count
        self.is_locked = (attempt_count >= 3)
        
        if self.is_locked:
            message = f"使用者 {username} 已被鎖定（失敗 {attempt_count} 次）"
        else:
            message = f"使用者 {username} 登入失敗（第 {attempt_count} 次）"
        
        super().__init__(message)

# 測試 1：第 1 次失敗（未鎖定）
try:
    raise InvalidCredentialsError("alice", 1)
except InvalidCredentialsError as e:
    print(f"例外訊息：{e}")
    print(f"是否鎖定：{e.is_locked}")

# 測試 2：第 3 次失敗（已鎖定）
try:
    raise InvalidCredentialsError("alice", 3)
except InvalidCredentialsError as e:
    print(f"\n例外訊息：{e}")
    print(f"是否鎖定：{e.is_locked}")
````

**重點提示**：
1. 在 `__init__` 中計算 `is_locked`：`attempt_count >= 3`
2. 使用 if-else 產生不同的錯誤訊息
3. 三個屬性都要儲存：`username`, `attempt_count`, `is_locked`

**常見錯誤**：
- ❌ 忘記計算 `is_locked` 屬性
- ❌ 條件判斷錯誤（使用 `>` 而非 `>=`）
- ❌ 訊息格式不符合規格

### 習題 3：divide_all 函式（10 分）

**題目重述**：實作 `divide_all(numbers, divisor)`，對列表中的每個數字執行除法。

**知識點**：
- ✅ 在函式開頭進行輸入驗證
- ✅ 使用 `raise` 拋出內建例外（`ValueError`）
- ✅ 列表推導式應用

````python
# 習題 3 完整解答

def divide_all(numbers, divisor):
    """
    對列表中的每個數字執行除法
    
    Args:
        numbers: 數字列表
        divisor: 除數
    
    Returns:
        list: 結果列表
    
    Raises:
        ValueError: 除數為 0 或列表為空時
    """
    # 驗證 1：檢查除數
    if divisor == 0:
        raise ValueError("除數不能為 0")
    
    # 驗證 2：檢查列表是否為空
    if not numbers:  # 等同於 len(numbers) == 0
        raise ValueError("列表不能為空")
    
    # 執行除法
    return [num / divisor for num in numbers]

# 測試 1：正常情況
result = divide_all([10, 20, 30], 5)
print(f"結果：{result}")  # [2.0, 4.0, 6.0]

# 測試 2：除數為 0
try:
    divide_all([10], 0)
except ValueError as e:
    print(f"錯誤：{e}")

# 測試 3：空列表
try:
    divide_all([], 5)
except ValueError as e:
    print(f"錯誤：{e}")
````

**重點提示**：
1. **驗證順序**：先檢查除數，再檢查列表（避免除以零優先於空列表）
2. **空列表檢查**：使用 `if not numbers` 比 `if len(numbers) == 0` 更簡潔
3. **列表推導式**：`[num / divisor for num in numbers]` 是最簡潔的寫法

**常見錯誤**：
- ❌ 驗證順序錯誤（先檢查列表再檢查除數）
- ❌ 使用 `raise ZeroDivisionError` 而非 `ValueError`
- ❌ 錯誤訊息與規格不符

**延伸思考**：
- 如果列表包含非數字元素怎麼辦？（會拋出 `TypeError`）
- 是否需要檢查 `numbers` 的型態？（視需求而定）

### 習題 4：OrderError 例外階層（15 分）

**題目重述**：為電商平台設計訂單處理的例外階層。

**知識點**：
- ✅ 三層例外階層設計
- ✅ 繼承關係（基底 → 分類 → 具體）
- ✅ 每個例外攜帶相關屬性
- ✅ 分層捕獲（從具體到抽象）

````python
# 習題 4 完整解答

# 第一層：基底例外
class OrderError(Exception):
    """訂單相關錯誤的基底類別"""
    pass

# 第二層：分類例外
class ProductError(OrderError):
    """產品相關錯誤"""
    pass

class PaymentError(OrderError):
    """付款相關錯誤"""
    pass

# 第三層：具體例外（產品類）
class ProductOutOfStockError(ProductError):
    """庫存不足"""
    def __init__(self, product_id, requested, available):
        self.product_id = product_id
        self.requested = requested
        self.available = available
        message = f"產品 {product_id} 庫存不足：需要 {requested}，只有 {available}"
        super().__init__(message)

class ProductNotFoundError(ProductError):
    """產品不存在"""
    def __init__(self, product_id):
        self.product_id = product_id
        message = f"產品 {product_id} 不存在"
        super().__init__(message)

# 第三層：具體例外（付款類）
class PaymentDeclinedError(PaymentError):
    """付款被拒絕"""
    def __init__(self, reason):
        self.reason = reason
        message = f"付款被拒絕：{reason}"
        super().__init__(message)

class PaymentTimeoutError(PaymentError):
    """付款逾時"""
    def __init__(self, timeout_seconds):
        self.timeout_seconds = timeout_seconds
        message = f"付款逾時：超過 {timeout_seconds} 秒"
        super().__init__(message)
````

````python
# 測試：分層捕獲示範

import random

def random_order_error():
    """隨機拋出不同的訂單錯誤"""
    choice = random.choice([1, 2, 3, 4])
    if choice == 1:
        raise ProductOutOfStockError("PROD001", requested=10, available=5)
    elif choice == 2:
        raise ProductNotFoundError("PROD999")
    elif choice == 3:
        raise PaymentDeclinedError("信用卡餘額不足")
    else:
        raise PaymentTimeoutError(30)

# 分層捕獲：從具體到抽象
try:
    random_order_error()
except ProductOutOfStockError as e:
    print(f"🔴 庫存不足：{e.product_id}，需要 {e.requested}，只有 {e.available}")
except ProductError as e:
    print(f"⚠️ 產品錯誤：{e}")
except PaymentError as e:
    print(f"💳 付款錯誤：{e}")
except OrderError as e:
    print(f"📦 訂單錯誤：{e}")
````

**重點提示**：
1. **階層結構**：`OrderError` → `ProductError/PaymentError` → 具體例外
2. **屬性設計**：每個具體例外攜帶相關資訊（如 `product_id`, `requested`, `available`）
3. **捕獲順序**：從具體到抽象（子類別在前，父類別在後）
4. **訊息格式**：清晰描述錯誤原因，包含關鍵資訊

**常見錯誤**：
- ❌ 繼承關係錯誤（如 `ProductOutOfStockError` 直接繼承 `OrderError`）
- ❌ 捕獲順序錯誤（父類別在前會吃掉子類別的例外）
- ❌ 忘記儲存屬性

### 習題 5：assert vs raise（10 分）

**題目重述**：將錯誤檢查改為適當的 `raise` 或 `assert`。

**知識點**：
- ✅ `assert` 用於內部假設（開發者錯誤）
- ✅ `raise` 用於外部驗證（使用者輸入錯誤）
- ✅ `assert` 可被停用（`python -O`）
- ✅ 選擇適當的例外類型（`TypeError`, `ValueError`）

````python
# 習題 5 完整解答

def process(data):
    """模擬的處理函式"""
    return [data['name']]

def process_user_input(data):
    """
    處理使用者輸入
    
    Args:
        data: 使用者資料（字典）
    
    Returns:
        list: 處理結果
    
    Raises:
        TypeError: data 不是字典時
        ValueError: data 缺少必要欄位時
    """
    # 檢查 1：data 不應為 None（內部假設）
    assert data is not None, "data 是 None（內部錯誤）"
    
    # 檢查 2：data 必須是字典（外部輸入驗證）
    if not isinstance(data, dict):
        raise TypeError(f"data 必須是字典，但收到 {type(data).__name__}")
    
    # 檢查 3：必須包含 'name' 欄位（外部輸入驗證）
    if 'name' not in data:
        raise ValueError("data 缺少必要欄位 'name'")
    
    result = process(data)
    
    # 檢查 4：內部變數 result 不應為空列表（內部假設）
    assert result != [], "result 不應為空列表（內部錯誤）"
    
    return result
````

````python
# 測試程式碼

# 測試 1：正常情況
result = process_user_input({'name': 'Alice'})
print(f"結果：{result}")

# 測試 2：不是字典
try:
    process_user_input("not a dict")
except TypeError as e:
    print(f"捕獲 TypeError：{e}")

# 測試 3：缺少欄位
try:
    process_user_input({'age': 25})
except ValueError as e:
    print(f"捕獲 ValueError：{e}")
````

**重點提示**：

**assert vs raise 選擇原則**：

| 情境 | 使用 | 原因 |
|:-----|:-----|:-----|
| data 不應為 None | `assert` | 內部假設，呼叫者應確保傳入非 None |
| data 必須是字典 | `raise TypeError` | 外部驗證，使用者可能傳錯型態 |
| data 缺少 'name' | `raise ValueError` | 外部驗證，使用者可能傳錯內容 |
| result 不應為空 | `assert` | 內部假設，`process()` 應回傳非空列表 |

**常見錯誤**：
- ❌ 用 `assert` 驗證使用者輸入（會被 `-O` 停用）
- ❌ 用 `raise` 檢查內部假設（過度防禦）
- ❌ 所有檢查都用 `raise ValueError`（不夠精確）

### 習題 6：safe_divide 例外鏈（10 分）

**題目重述**：實作 `safe_divide(a, b)`，展示如何捕獲、記錄、修改並重新拋出例外。

**知識點**：
- ✅ `raise ... from e` 建立例外鏈
- ✅ 保留原始 traceback
- ✅ 在重新拋出前記錄日誌
- ✅ `__cause__` 屬性存取

````python
# 習題 6 完整解答

def safe_divide(a, b):
    """
    安全的除法運算
    
    Args:
        a: 被除數
        b: 除數
    
    Returns:
        float: 商
    
    Raises:
        ValueError: 除數為 0 時
    """
    try:
        return a / b
    except ZeroDivisionError as e:
        # 記錄日誌
        print(f"[LOG] 除以零錯誤：a={a}, b={b}")
        
        # 拋出更友善的錯誤訊息，並保留原始例外
        raise ValueError("無法執行除法：除數為 0") from e

# 測試 1：正常除法
result = safe_divide(10, 2)
print(f"10 / 2 = {result}")

# 測試 2：除以零
try:
    safe_divide(10, 0)
except ValueError as e:
    print(f"\n捕獲 ValueError：{e}")
    print(f"原始錯誤（__cause__）：{e.__cause__}")
    print(f"原始錯誤類型：{type(e.__cause__).__name__}")
````

**重點提示**：
1. **例外鏈語法**：`raise NewError(...) from original_error`
2. **`__cause__` 屬性**：顯式例外鏈，由 `from` 設定
3. **記錄日誌**：在 `except` 區塊中處理，再重新拋出
4. **更友善的錯誤訊息**：`ValueError("無法執行除法：除數為 0")` 比 `ZeroDivisionError` 更清楚

**例外鏈的優勢**：
- ✅ 保留完整的 traceback 資訊
- ✅ 顯示錯誤的因果關係
- ✅ 方便除錯和診斷

**常見錯誤**：
- ❌ 使用 `raise ValueError(...)` 而非 `raise ... from e`（丟失原始錯誤）
- ❌ 在 `except` 區塊中使用裸 `raise`（會直接拋出 `ZeroDivisionError`）
- ❌ 忘記記錄日誌

### 習題 7：多層例外鏈追蹤（15 分）

**題目重述**：實作多層資料處理系統，展示例外鏈的完整追蹤。

**知識點**：
- ✅ 三層處理流程（fetch → parse → validate）
- ✅ 每層使用 `raise ... from` 包裝例外
- ✅ 使用 `while` 迴圈追蹤 `__cause__` 鏈
- ✅ 模擬真實的錯誤處理流程

````python
# 習題 7 完整解答

import json
import random

# 定義例外階層
class DataError(Exception):
    """資料處理錯誤基底類別"""
    pass

class NetworkError(DataError):
    """網路錯誤"""
    pass

class ParseError(DataError):
    """解析錯誤"""
    pass

class ValidationError(DataError):
    """驗證錯誤"""
    pass

# 第一層：fetch_data
def fetch_data(url):
    """模擬網路請求"""
    try:
        # 模擬 30% 機率網路失敗
        if random.random() < 0.3:
            raise ConnectionError(f"無法連接到 {url}")
        
        # 模擬返回 JSON 字串（故意不完整以測試解析錯誤）
        if random.random() < 0.3:
            return "{invalid json"  # 無效 JSON
        
        # 正常情況：返回完整 JSON
        return '{"id": 123, "name": "Alice"}'
    
    except ConnectionError as e:
        raise NetworkError(f"網路請求失敗：{url}") from e

# 第二層：parse_data
def parse_data(text):
    """解析 JSON"""
    try:
        return json.loads(text)
    except json.JSONDecodeError as e:
        raise ParseError(f"無法解析 JSON：{e}") from e

# 第三層：validate_data
def validate_data(data):
    """驗證資料"""
    if 'id' not in data:
        raise ValidationError("資料缺少必要欄位 'id'")
    return data

# 完整流程
def process_pipeline(url):
    """完整的處理流程"""
    text = fetch_data(url)          # 可能拋出 NetworkError
    data = parse_data(text)         # 可能拋出 ParseError
    validated = validate_data(data) # 可能拋出 ValidationError
    return validated
````

````python
# 測試：例外鏈追蹤

try:
    result = process_pipeline("https://api.example.com/data")
    print(f"處理成功：{result}")
except DataError as e:
    print(f"\n❌ 資料錯誤：{e}")
    
    # 追蹤例外鏈
    print("\n例外鏈追蹤：")
    current = e
    level = 0
    while current is not None:
        indent = "  " * level
        print(f"{indent}Level {level}: {type(current).__name__}")
        print(f"{indent}  訊息: {current}")
        current = current.__cause__
        level += 1
    
    print(f"\n總共追蹤了 {level} 層例外")
````

**重點提示**：
1. **三層架構**：
   - `fetch_data`: 捕獲 `ConnectionError` → 包裝成 `NetworkError`
   - `parse_data`: 捕獲 `JSONDecodeError` → 包裝成 `ParseError`
   - `validate_data`: 直接拋出 `ValidationError`

2. **追蹤例外鏈**：
   ```python
   current = e
   while current is not None:
       print(type(current).__name__, current)
       current = current.__cause__
   ```

3. **可能的例外鏈**：
   - 網路失敗：`NetworkError` → `ConnectionError`
   - 解析失敗：`ParseError` → `JSONDecodeError`
   - 驗證失敗：`ValidationError` （無鏈）

**常見錯誤**：
- ❌ 忘記使用 `from` 建立例外鏈
- ❌ 追蹤例外鏈時忘記檢查 `current is not None`（會無限迴圈）
- ❌ 所有層都直接拋出，沒有包裝（失去抽象化）

### 習題 8：raise from None 抑制鏈（10 分）

**題目重述**：使用 `raise ... from None` 抑制例外鏈，隱藏內部實作細節。

**知識點**：
- ✅ `from None` 抑制例外鏈
- ✅ `__cause__` 和 `__context__` 都為 None
- ✅ 適用於公開 API（不暴露內部錯誤）

````python
# 習題 8 完整解答

class DatabaseError(Exception):
    """內部資料庫錯誤（不應暴露給使用者）"""
    pass

class UserNotFoundError(Exception):
    """公開 API 例外：使用者不存在"""
    pass

def _db_fetch(user_id):
    """模擬資料庫查詢（內部函式）"""
    if user_id == "bad_id":
        raise DatabaseError(f"Database connection failed for {user_id}")
    return {"id": user_id, "name": "Alice"}

def api_get_user(user_id):
    """
    取得使用者資料（公開 API）
    
    Args:
        user_id: 使用者 ID
    
    Returns:
        dict: 使用者資料
    
    Raises:
        UserNotFoundError: 使用者不存在時
    """
    try:
        return _db_fetch(user_id)
    except DatabaseError:
        # 使用 from None 隱藏內部 DatabaseError
        raise UserNotFoundError(f"使用者 {user_id} 不存在") from None
````

````python
# 測試程式碼

# 測試 1：正常情況
user = api_get_user("user123")
print(f"使用者：{user}")

# 測試 2：使用者不存在
try:
    api_get_user("bad_id")
except UserNotFoundError as e:
    print(f"\n捕獲例外：{e}")
    print(f"__cause__: {e.__cause__}")      # None
    print(f"__context__: {e.__context__}")  # None
    print("\n✅ 成功隱藏內部 DatabaseError")
````

**重點提示**：

**`from None` 的作用**：
1. 設定 `__cause__ = None`（顯式例外鏈）
2. 設定 `__suppress_context__ = True`（隱藏隱式例外鏈）
3. 結果：`__cause__` 和 `__context__` 都為 None

**使用時機**：
- ✅ 公開 API：不想暴露內部實作細節
- ✅ 安全考量：避免洩漏敏感資訊（如資料庫結構）
- ✅ 使用者友善：提供更清晰的錯誤訊息

**對比**：

| 語法 | __cause__ | __context__ | 用途 |
|:-----|:----------|:------------|:-----|
| `raise NewError()` | None | 原始錯誤 | 隱式鏈（自動） |
| `raise NewError() from e` | 原始錯誤 | 原始錯誤 | 顯式鏈（手動） |
| `raise NewError() from None` | None | None | 抑制鏈（隱藏） |

**常見錯誤**：
- ❌ 濫用 `from None`（過度隱藏資訊，難以除錯）
- ❌ 在內部函式使用（應該在 API 邊界使用）
- ❌ 忘記捕獲內部例外就直接用 `from None`

### 習題 9：完整的 ATM 系統（15 分）

**題目重述**：設計並實作一個 ATM 提款系統的完整例外處理。

**知識點**：
- ✅ 三層例外階層（9 個例外類別）
- ✅ 依序驗證流程
- ✅ 每個例外攜帶詳細資訊
- ✅ 分層捕獲示範

````python
# 習題 9 完整解答（第一部分：例外階層）

# 第一層：基底例外
class ATMError(Exception):
    """ATM 錯誤基底類別"""
    pass

# 第二層：分類例外
class CardError(ATMError):
    """卡片相關錯誤"""
    pass

class PINError(ATMError):
    """PIN 相關錯誤"""
    pass

class TransactionError(ATMError):
    """交易相關錯誤"""
    pass

# 第三層：具體例外（卡片類）
class InvalidCardError(CardError):
    """卡號無效"""
    def __init__(self, card_number):
        self.card_number = card_number
        super().__init__(f"卡號無效：{card_number}")

class CardExpiredError(CardError):
    """卡片過期"""
    def __init__(self, card_number, expiry_date):
        self.card_number = card_number
        self.expiry_date = expiry_date
        super().__init__(f"卡片 {card_number} 已於 {expiry_date} 過期")

class CardBlockedError(CardError):
    """卡片被鎖定"""
    def __init__(self, card_number):
        self.card_number = card_number
        super().__init__(f"卡片 {card_number} 已被鎖定")

# 第三層：具體例外（PIN 類）
class InvalidPINError(PINError):
    """PIN 錯誤"""
    def __init__(self):
        super().__init__("PIN 碼錯誤")

class PINAttemptsExceededError(PINError):
    """超過 PIN 嘗試次數"""
    def __init__(self, attempts):
        self.attempts = attempts
        super().__init__(f"PIN 輸入錯誤超過 {attempts} 次")

# 第三層：具體例外（交易類）
class InsufficientFundsError(TransactionError):
    """餘額不足"""
    def __init__(self, amount, balance):
        self.amount = amount
        self.balance = balance
        super().__init__(f"餘額不足：需要 ${amount}，但只有 ${balance}")

class DailyLimitExceededError(TransactionError):
    """超過每日限額"""
    def __init__(self, amount, limit, used_today):
        self.amount = amount
        self.limit = limit
        self.used_today = used_today
        remaining = limit - used_today
        super().__init__(
            f"超過每日限額：需要 ${amount}，但今日剩餘額度 ${remaining}"
        )

class ATMOutOfCashError(TransactionError):
    """ATM 現金不足"""
    def __init__(self, requested, available):
        self.requested = requested
        self.available = available
        super().__init__(
            f"ATM 現金不足：需要 ${requested}，ATM 只有 ${available}"
        )
````

````python
# 習題 9 完整解答（第二部分：ATM 類別）

class ATM:
    def __init__(self, cash_available=100000):
        self.cash_available = cash_available
    
    def withdraw(self, card_number, pin, amount, account_balance, 
                 daily_limit, used_today):
        """
        執行提款
        
        Args:
            card_number: 卡號（字串）
            pin: PIN 碼（字串）
            amount: 提款金額
            account_balance: 帳戶餘額
            daily_limit: 每日限額
            used_today: 今日已提款金額
        
        Returns:
            dict: 交易資訊
        
        Raises:
            ATMError: 各種 ATM 錯誤
        """
        # 驗證 1：檢查卡號格式（16 位數字）
        if not (card_number.isdigit() and len(card_number) == 16):
            raise InvalidCardError(card_number)
        
        # 驗證 2：檢查 PIN（4 位數字）
        if not (pin.isdigit() and len(pin) == 4):
            raise InvalidPINError()
        
        # 驗證 3：檢查帳戶餘額
        if amount > account_balance:
            raise InsufficientFundsError(amount, account_balance)
        
        # 驗證 4：檢查每日限額
        if used_today + amount > daily_limit:
            raise DailyLimitExceededError(amount, daily_limit, used_today)
        
        # 驗證 5：檢查 ATM 現金
        if amount > self.cash_available:
            raise ATMOutOfCashError(amount, self.cash_available)
        
        # 所有驗證通過，執行提款
        self.cash_available -= amount
        
        return {
            "success": True,
            "amount": amount,
            "remaining_balance": account_balance - amount,
            "atm_cash_remaining": self.cash_available
        }
````

````python
# 測試程式碼

atm = ATM(cash_available=50000)

# 測試 1：正常提款
try:
    result = atm.withdraw(
        card_number="1234567890123456",
        pin="1234",
        amount=5000,
        account_balance=10000,
        daily_limit=20000,
        used_today=0
    )
    print(f"✅ 提款成功：{result}")
except ATMError as e:
    print(f"❌ 錯誤：{e}")

# 測試 2：卡號無效
try:
    atm.withdraw("123", "1234", 1000, 10000, 20000, 0)
except InvalidCardError as e:
    print(f"\n❌ 卡號錯誤：{e}")
except ATMError as e:
    print(f"❌ 其他錯誤：{e}")

# 測試 3：餘額不足
try:
    atm.withdraw("1234567890123456", "1234", 15000, 10000, 20000, 0)
except InsufficientFundsError as e:
    print(f"\n❌ {e}")
    print(f"   需要：${e.amount}，餘額：${e.balance}")
except ATMError as e:
    print(f"❌ 其他錯誤：{e}")

# 測試 4：ATM 現金不足
try:
    atm.withdraw("1234567890123456", "1234", 60000, 100000, 100000, 0)
except ATMOutOfCashError as e:
    print(f"\n❌ {e}")
    print(f"   需要：${e.requested}，ATM 剩餘：${e.available}")
except ATMError as e:
    print(f"❌ 其他錯誤：{e}")
````

**重點提示**：

**1. 例外階層設計**：
```
ATMError (基底)
├── CardError (分類)
│   ├── InvalidCardError
│   ├── CardExpiredError
│   └── CardBlockedError
├── PINError (分類)
│   ├── InvalidPINError
│   └── PINAttemptsExceededError
└── TransactionError (分類)
    ├── InsufficientFundsError
    ├── DailyLimitExceededError
    └── ATMOutOfCashError
```

**2. 驗證順序**：
1. 卡號格式 → 2. PIN 格式 → 3. 餘額 → 4. 每日限額 → 5. ATM 現金

**3. 屬性設計**：
- 每個例外攜帶相關資訊（如 `amount`, `balance`）
- 錯誤訊息包含關鍵數值
- 方便後續處理和記錄

**4. 分層捕獲優勢**：
```python
except InvalidCardError:     # 特定處理
except CardError:            # 分類處理
except ATMError:             # 通用處理
```

### 習題 10：例外處理最佳實踐（5 分）

**題目重述**：找出並修正程式碼中的例外處理問題。

**知識點**：
- ✅ 避免裸 `except`
- ✅ 正確使用 `assert`
- ✅ 使用 `with` 管理檔案
- ✅ 避免吞沒例外
- ✅ 記錄錯誤訊息

````python
# 習題 10 完整解答

def parse(data):
    """模擬的解析函式"""
    return [x.strip() for x in data.split(',')]

# === 原始程式碼（有問題）===

def process_file_bad(filename):
    # 問題 1：裸 except（不知道捕獲什麼錯誤）
    # 問題 2：檔案未正確關閉（如果 parse 失敗）
    try:
        f = open(filename, 'r')
        data = f.read()
        f.close()
        result = parse(data)
        return result
    except:
        print("發生錯誤")
        return None

def calculate_bad(data):
    # 問題 3：使用 assert 驗證外部輸入（應該用 raise）
    # 問題 4：assert 可被 -O 停用
    assert data is not None
    assert len(data) > 0
    
    total = 0
    for item in data:
        # 問題 5：裸 except 吞沒所有錯誤
        try:
            value = int(item)
        except:
            value = 0
        total += value
    return total / len(data)
````

````python
# === 修正後的程式碼 ===

def process_file(filename):
    """
    處理檔案並解析內容
    
    修正：
    1. 使用 with 語句確保檔案正確關閉
    2. 捕獲具體例外（FileNotFoundError, IOError）
    3. 記錄錯誤訊息（使用 logging 更佳）
    4. 拋出例外而非返回 None
    """
    try:
        # 修正 1：使用 with 語句
        with open(filename, 'r') as f:
            data = f.read()
        result = parse(data)
        return result
    # 修正 2：捕獲具體例外
    except FileNotFoundError as e:
        print(f"[ERROR] 檔案不存在：{filename}")
        raise  # 重新拋出，不吞沒例外
    except IOError as e:
        print(f"[ERROR] 讀取檔案失敗：{e}")
        raise

def calculate(data):
    """
    計算平均值
    
    修正：
    3. 使用 raise 驗證外部輸入（而非 assert）
    4. 捕獲具體例外（ValueError）
    5. 記錄無效資料而非靜默忽略
    """
    # 修正 3 & 4：使用 raise 而非 assert
    if data is None:
        raise ValueError("data 不能為 None")
    if len(data) == 0:
        raise ValueError("data 不能為空列表")
    
    total = 0
    valid_count = 0
    
    for item in data:
        # 修正 5：捕獲具體例外，記錄無效資料
        try:
            value = int(item)
            total += value
            valid_count += 1
        except ValueError:
            print(f"[WARNING] 無法轉換為整數：'{item}'，跳過")
            # 不再靜默設為 0，而是跳過
    
    if valid_count == 0:
        raise ValueError("沒有有效的數字資料")
    
    return total / valid_count
````

````python
# 測試修正後的程式碼

# 建立測試檔案
with open("test_data.txt", "w") as f:
    f.write("10, 20, 30, abc, 40")

try:
    data = process_file("test_data.txt")
    print(f"資料：{data}")
    
    result = calculate(data)
    print(f"平均值：{result}")
    
except (FileNotFoundError, IOError, ValueError) as e:
    print(f"錯誤：{type(e).__name__} - {e}")
````

**問題總結**：

| 問題編號 | 原始問題 | 修正方式 |
|:--------|:---------|:---------|
| 問題 1 | 裸 `except` | 捕獲具體例外（`FileNotFoundError`, `IOError`） |
| 問題 2 | 檔案未正確關閉 | 使用 `with` 語句 |
| 問題 3 | 用 `assert` 驗證輸入 | 改用 `raise ValueError` |
| 問題 4 | `assert` 可被停用 | 使用 `if ... raise` |
| 問題 5 | 裸 `except` 吞沒錯誤 | 捕獲 `ValueError`，記錄警告 |

**最佳實踐總結**：

1. **避免裸 `except`**：
   ```python
   # ❌ 錯誤
   except:
       pass
   
   # ✅ 正確
   except ValueError as e:
       print(f"錯誤：{e}")
   ```

2. **使用 `with` 管理資源**：
   ```python
   # ❌ 錯誤
   f = open(file)
   data = f.read()
   f.close()
   
   # ✅ 正確
   with open(file) as f:
       data = f.read()
   ```

3. **`assert` vs `raise`**：
   ```python
   # ❌ 錯誤（驗證外部輸入）
   assert data is not None
   
   # ✅ 正確
   if data is None:
       raise ValueError("data 不能為 None")
   ```

4. **不要吞沒例外**：
   ```python
   # ❌ 錯誤
   except:
       return None
   
   # ✅ 正確
   except ValueError as e:
       print(f"錯誤：{e}")
       raise  # 重新拋出
   ```

5. **記錄錯誤訊息**：
   ```python
   # ❌ 錯誤
   print("發生錯誤")
   
   # ✅ 正確
   print(f"[ERROR] 檔案不存在：{filename}")
   # 或使用 logging
   logging.error(f"檔案不存在：{filename}")
   ```

### 核心概念回顧

| 概念 | 關鍵習題 | 核心要點 |

| from None | 8 | 抑制例外鏈，隱藏內部錯誤 |
| 最佳實踐 | 10 | 避免裸 except，使用 with，記錄錯誤 |

### 例外處理決策樹

```
需要拋出例外？
├─ 是內部假設
│   └─ 使用 assert（開發階段檢查）
└─ 是外部驗證
    ├─ 內建例外足夠？
    │   └─ 是 → 使用內建例外（ValueError, TypeError...）
    └─ 否 → 建立自訂例外
        ├─ 單一錯誤 → 簡單例外類別
        └─ 多種相關錯誤 → 設計例外階層

需要重新拋出例外？
├─ 保留原始錯誤 → raise ... from e
├─ 隱藏原始錯誤 → raise ... from None
└─ 直接拋出 → raise（在 except 區塊中）
```

### 常見模式總結

**模式 1：自訂例外模板**
```python
class MyError(BaseError):
    def __init__(self, arg1, arg2):
        self.arg1 = arg1
        self.arg2 = arg2
        message = f"錯誤：{arg1}, {arg2}"
        super().__init__(message)
```

**模式 2：例外鏈**
```python
try:
    low_level_operation()
except LowError as e:
    raise HighError("高階錯誤") from e
```

**模式 3：分層捕獲**
```python
try:
    operation()
except SpecificError:    # 特定處理
    handle_specific()
except CategoryError:    # 分類處理
    handle_category()
except BaseError:        # 通用處理
    handle_base()
```
