### Part IV: 程式填空題參考答案

#### Q26 解答

````python
text = "hello world"
frequency = {}

for char in text:
    if char != " ":  # 忽略空格
        frequency[char] = frequency.get(char, 0) + 1

print(frequency)
# 輸出: {'h': 1, 'e': 1, 'l': 3, 'o': 2, 'w': 1, 'r': 1, 'd': 1}
````

**替代方案（字典推導式）**：

````python
text = "hello world"
chars = [c for c in text if c != " "]
frequency = {char: chars.count(char) for char in set(chars)}
print(frequency)
````

**最佳方案（使用 Counter）**：

````python
from collections import Counter
text = "hello world"
frequency = Counter(text.replace(" ", ""))
print(dict(frequency))
````

#### Q27 解答

````python
scores = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 58,
    "David": 95,
    "Eve": 73
}

# 1. 計算平均分數
average = sum(scores.values()) / len(scores)
print(f"平均分數: {average:.2f}")

# 2. 找出最高分學生
max_score = max(scores.values())
top_student = [name for name, score in scores.items() if score == max_score][0]
print(f"最高分學生: {top_student} ({max_score} 分)")

# 3. 篩選及格學生
passing = [name for name, score in scores.items() if score >= 60]
print(f"及格學生: {passing}")

# 輸出:
# 平均分數: 80.60
# 最高分學生: David (95 分)
# 及格學生: ['Alice', 'Bob', 'David', 'Eve']
````

#### Q28 解答

````python
en_to_zh = {
    "apple": "蘋果",
    "banana": "香蕉",
    "cherry": "櫻桃",
    "date": "棗子"
}

# 1. 建立中英翻譯字典（鍵值對調）
zh_to_en = {zh: en for en, zh in en_to_zh.items()}
print(f"中英字典: {zh_to_en}")

# 2. 篩選長度 > 5 的英文單字
long_words = {en: zh for en, zh in en_to_zh.items() if len(en) > 5}
print(f"長單字: {long_words}")

# 輸出:
# 中英字典: {'蘋果': 'apple', '香蕉': 'banana', '櫻桃': 'cherry', '棗子': 'date'}
# 長單字: {'banana': '香蕉', 'cherry': '櫻桃'}
````
