## 🔑 重點概念複習

### 四種推導式語法對照

```python
# 列表推導式（最常用）
[expression for item in iterable]
[expression for item in iterable if condition]

# 字典推導式（需要 key: value）
{key: value for item in iterable}
{key: value for item in iterable if condition}

# 集合推導式（自動去重）
{expression for item in iterable}
{expression for item in iterable if condition}

# 生成器表達式（惰性求值）
(expression for item in iterable)
(expression for item in iterable if condition)
```

### 條件的兩種用法

```python
# 1. 篩選條件（if 在後）
[x for x in range(10) if x % 2 == 0]  # [0, 2, 4, 6, 8]

# 2. 條件表達式（if-else 在前）
[x if x % 2 == 0 else 0 for x in range(10)]  # [0, 0, 2, 0, 4, 0, 6, 0, 8, 0]
```

### 生成器 vs 列表

```python
# 列表推導式（立即建立所有元素）
lst = [x for x in range(1000000)]  # 記憶體使用大
list(lst)  # 可多次迭代

# 生成器表達式（惰性求值）
gen = (x for x in range(1000000))  # 記憶體使用小
list(gen)  # 只能迭代一次
```
